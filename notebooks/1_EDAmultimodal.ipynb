# Suppression de la classe 1

```python
def remove_specified_classes(df, label_column, class_to_remove=1):
    """
    Cette fonction supprime les classes spécifiées du DataFrame df.
    """
    df_filtered = df[~df[label_column].str.match(rf'^{class_to_remove}$')]
    
    # Supprimez les classes spécifiées
    df_filtered[label_column] = df_filtered[label_column].apply(
        lambda x: ' '.join([label for label in x.split() if label != str(class_to_remove)])
    )
    
    return df_filtered
```

```python
df_filtered = remove_specified_classes(df, 'Labels')
df_filtered["Labels_list"] = df_filtered["Labels"].apply(lambda x: list(map(int, x.split())))
df_filtered.head()
```

```python
# Extraire tous les labels dans une liste commune
all_labels = list(chain.from_iterable(df_filtered["Labels_list"]))

# Extraire tous les labels dans une liste commune
all_labels = list(chain.from_iterable(df_filtered["Labels_list"]))

# Compter la fréquence de chaque label
label_counts = Counter(all_labels)

```

```python
# Afficher la distribution individuelle
print("Distribution de chaque label :")
for label, freq in label_counts.most_common():
    print(f"Label {label} : {freq} occurrences")

labels_unique = sorted(label_counts.keys())
frequencies = [label_counts[l] for l in labels_unique]

plt.figure(figsize=(10, 5))
plt.bar(labels_unique, frequencies)
plt.xlabel("Label")
plt.ylabel("Nombre d'occurrences")
plt.title("Distribution de chaque label dans le dataset")
plt.show()
```

```python
import shutil

def copy_images(csv_path, dest):
    # Lire le fichier CSV avec pandas
    df = pd.read_csv(csv_path, on_bad_lines='warn')
    
    # Créer le dossier de destination s'il n'existe pas
    if not os.path.exists(dest):
        os.makedirs(dest)
        
    # Compteurs pour le rapport
    copied = 0
    not_found = 0
    
    for image_id in df['ImageID']:
        image_path = os.path.join(image_folder, image_id)
        if os.path.exists(image_path):
            shutil.copy(image_path, dest)
            copied += 1
        else:
            print(f"Image non trouvée: {image_id}")
            not_found += 1
    
    print(f"Copie terminée pour {dest}: {copied} images copiées, {not_found} images non trouvées")

```

```python
# Appel de la fonction avec des chemins de CSV
train_data = 'C:/Users/Dell/OneDrive/Desktop/Projet_final_BIHAR/Jean_Duckens_Final_Evaluation/data/processed/train_split.csv'
val_data = 'C:/Users/Dell/OneDrive/Desktop/Projet_final_BIHAR/Jean_Duckens_Final_Evaluation/data/processed/val_split.csv'
test_data = 'C:/Users/Dell/OneDrive/Desktop/Projet_final_BIHAR/Jean_Duckens_Final_Evaluation/data/processed/test_split.csv'

# Copier les images
copy_images(train_data, 'C:/Users/Dell/OneDrive/Desktop/Projet_final_BIHAR/Jean_Duckens_Final_Evaluation/data/train_image')
copy_images(val_data, 'C:/Users/Dell/OneDrive/Desktop/Projet_final_BIHAR/Jean_Duckens_Final_Evaluation/data/val_image')
copy_images(test_data, 'C:/Users/Dell/OneDrive/Desktop/Projet_final_BIHAR/Jean_Duckens_Final_Evaluation/data/test_image')
```
